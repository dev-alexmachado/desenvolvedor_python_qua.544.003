Hoje, iremos aprender a trabalhar com células Python. As células Python executam códigos Python dentro dos notebooks. Observe a célula Python abaixo:

```python
nome = "Alex Machado"
texto = f"Esta é uma célula Python escrita pelo melhor programador de todos os tempos: {nome} 😎"

print(texto)
```

A célula Python também permite códigos de entrada de dados. Veja:

```python
nome = input("Informe seu nome: ").strip().title()
print(nome)
```

As células Python são acumulativas, e reconhecem as variáveis das células anteriores.

Observe, por exemplo, a célula abaixo:

```python
n1 = 5
print(f"O valor da variável 'n1' é {n1}.")
```

O valor de n1 é claramente 5. Porém, veja na próxima celula e diga qual será o valor da variável result:

```python
result = 15 + n1
print(result)
```

O valor de result é 20, pois está somando 15 com n1, que tinha o valor de 5.

Mas agora observe a célula abaixo sem executá-la:

```python
x = 10
```

Agora, faça o código abaixo na próxima célula, execute e veja qual é o resultado:

```python
result = 25 + x
print(result)
```
